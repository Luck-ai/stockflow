```python
import pandas as pd
import re
from   lib.etl import SpreadSheet
from   sqlalchemy import create_engine
```

```python
def connect_sql_engine(user, password, host='postgres', port=5432, database='db'):
    return create_engine(f"postgresql://{user}:{password}@{host}:{str(port)}/{database}")

sql_engine = connect_sql_engine(
    "admin",
    "admin"
)
```

```python
sales = SpreadSheet({
    "fpath": "./assets/sales/ยอดขาย 01_2025.1.xlsx",
    "head_offset": 1,
    "head_cut": 1,
    "tail_cut": 3
}).read()

sales.get_df()
```

```python
def rename_entry (df):
    df.rename(
        columns = {
            'วันที่ทำรายการ': 'transaction_date',
            'รหัสสินค้า': 'item_id',
            'ช่องทางการขาย': 'platform',
            'จำนวน': 'quantity'
        },
        inplace = True
    )
    return df

def reformat_date (df):
    df['transaction_date'] = pd.to_datetime(
        df['transaction_date'],
        format="%d/%m/%Y"
    )
    return df

filtered = sales.extract_columns([
    'วันที่ทำรายการ',
    'รหัสสินค้า',
    'ช่องทางการขาย',
    'จำนวน'
]).df_opr(lambda df: df.sort_values(
    by='วันที่ทำรายการ', ascending=True, ignore_index=True
)).df_opr(rename_entry).df_opr(reformat_date)

# filtered.export_sql(sql_engine, "sales")
filtered.export_csv(f"./processed/{filtered.fname}.csv").get_df()
```

```python
skus = SpreadSheet({
    "fpath": "./assets/skus/หมวดหมู่สินค้า.xlsx",
    "head_offset": 1,
    "head_cut": 1,
    "tail_cut": 3,
}).read()

skus.get_df()
```

```python
def normalize_head (df):
    df.rename(
        columns = {
            'ชื่อสินค้า': 'name',
            'รหัสสินค้า': 'raw_id',
            'หมวดหมู่': 'category',
            'หมวดหมู่ย่อย': 'subcategory',
            'หน่วยสินค้า': 'unit',
        },
        inplace = True,
    )
    return df

def normalize_body (df):
    df['unit'] = (
        df['unit']
        .astype(str)
        .str.replace('ชิ้น', 'piece', regex=False)
        .str.replace('เซต', 'set', regex=False)
        .str.replace('ตัว', 'one-unit', regex=False)
        .str.replace('แพ็ค', 'pack', regex=False)
        .str.replace('คู่', 'pairs', regex=False)
        .str.replace('ม้วน', 'roll', regex=False)
        .str.replace('กล่อง', 'box', regex=False)
        .str.replace('รายการ', 'order', regex=False)
        .str.replace('ก้อน', 'cell/lump', regex=False)
        .str.replace('กิโล', 'kg', regex=False)
        .str.replace('แท่ง', 'piece/stick', regex=False)
    )
    df['size'] = (
        df['size']
        .astype(str)
        .str.replace('แบบหนา', 'thick', regex=False)
        .str.replace('แบบบาง', 'thin', regex=False)
        .str.replace('นิ้ว', 'inchs', regex=False)
    )
    return df

def separate(row):
    raw_id = str(row["raw_id"])
    parts = raw_id.rsplit("-", 1)   # split from the right, at most once

    white_lst = ["M", "L", "XL", "XXL", "3XL", "Free size"]
    black_lst = ["TTB", "OF", "WH"]

    row["size"] = None
    if len(parts) == 2 and parts[1] in white_lst:
        row["size"] = parts[1]
        return row
    if len(parts) < 2 or parts[0] in black_lst:
        return row
    # Try extracting from name (last parenthesis group)
    matches = re.findall(r"\((.*?)\)", str(row["name"]))
    if matches:
        row["size"] = matches[-1]   # last match
    return row

filtered = skus.extract_columns([
    'รหัสสินค้า',
    'ชื่อสินค้า',
    'หมวดหมู่',
    'หมวดหมู่ย่อย',
    'หน่วยสินค้า'
])\
.df_opr(lambda df: df.sort_values(by='ชื่อสินค้า', ascending=True, ignore_index=True))\
.df_opr(normalize_head)\
.df_opr(lambda df: df.apply(separate, axis=1))\
.df_opr(normalize_body)

# filtered.export_sql(sql_engine, "skus")
filtered.export_csv(f"./processed/{filtered.fname}.csv").get_df()
```

```python
stocks = SpreadSheet({
    "fpath": "./assets/stocks/สินค้า 7.xlsx",
    "head_offset": 1,
    "head_cut": 1,
    "tail_cut": 3,
}).read()

stocks.get_df()
```

```python
def normalize_head (df):
    df.rename(
        columns = {
            'รหัส': 'raw_id',
            'สินค้า': 'name',
            'หมวดหมู่': 'category',
            'จำนวน': 'quantity'
        },
        inplace = True,
    )
    return df

filtered = stocks.extract_columns([
    'รหัส',
    'สินค้า',
    'หมวดหมู่',
    'จำนวน'
])\
.df_opr(lambda df: df.sort_values(by='สินค้า', ascending=True, ignore_index=True))\
.df_opr(normalize_head)

# filtered.export_sql(sql_engine, "stock")
filtered.export_csv(f"./processed/{filtered.fname}.csv").get_df()
```
